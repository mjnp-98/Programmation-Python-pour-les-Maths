## <center>Exercice sur les listes</center>

Une liste est une structure permettant de stocker des éléments hétérogènes :

```python
list0 = [0, 5.4, "chaîne", True]
list0
```

Les listes sont mutables c'est-à-dire qu'il est possible de modifier un élément, d'en rajouter un ou d'en supprimer un sans avoir à redéfinir la liste.

```python
list0[3] = False
list0.append("nouveau")
list0.insert(2, 34)
list0.remove(0)
list0
```

En particulier, il faut faire attention en copiant une liste. Si l'on exécute le code suivant:

```python
list1 = list0
list1[2] = "change"
list0
```

alors **list0** est également modifiée et est égale à **list1**. En effet, la première instruction écrite dans la cellule ci-dessus ne créée pas un nouvel objet (dans la mémoire vive de l'ordinateur) mais créée simplement une autre variable pointant vers le même objet. Pour le vérifier, il est possible d'utiliser la commande **id** qui renvoie l'identifiant unique d'une variable : si deux variables ont le même identifiant, alors elles pointent vers le même objet. C'est bel et bien le cas ici :

```python
id(list0) == id(list1)
```

Pour créer une copie, il faut utiliser la méthode **copy** :

```python
list2 = list0.copy()
list2[2] = "rechange"
list0
```

qui ne modifie pas **list0**. Il est possible de modifier les éléments d'une liste à l'intérieur d'une fonction :

```python
def f(l):
    l[0] = 0
f(list0)
list0
```

Enfin, il est possible de créer des listes à partir d'une boucle :

```python
list3 = [2*i + 1 for i in range(10)]
list3
```

### Question 4 : Ecrire une fonction qui retourne la suite de Fibonacci

```python
def fibonacci(N):
    """Calcul des N premiers termes de la suite de Fibonacci avec N >= 2"""
    fib = [0, 1]
    for n in range(N - 2):
        fib.append(fib[n] + fib[n + 1]) # fib = fib + [fib[n] + fib[n + 1]]
    return fib
fibonacci(6)
```

### Question 5 : Ecrire une fonction qui retourne une ligne donnée du triangle de Pascal

```python
def pascal(N):
    """Donne la N-ème ligne du triangle de Pascal"""
    pas = [1]
    for n in range(N - 1):
        pas_n = pas + [1]
        for _ in range(len(pas) - 1):
            pas_n[_ + 1] = pas[_] + pas[_ + 1]
        pas = pas_n
    return pas    
pascal(6)
```

### Question 6 : Trouver la valeur de deux suites à la même itération

```python
def iteration(N, init_u, init_v):
    u, v = init_u, init_v
    for _ in range(N):
        temp = u
        u = u + v
        v = 2 * temp - v
    return [u, v]
[u, v] = iteration(100, 1, 1)
[u, v]
```

### Question 7 : Ecrire une fonction avec des conditions

```python
def vk(n0, K):
    v = [n0] * (K + 1)
    for k in range(K):
        if v[k] % 2 == 1:
            v[k + 1] = 3 * v[k] + 1
        else:
            v[k + 1] = .5 * v[k]
    return v[len(v) - 5:]
K = 1000
print(vk(10, K) == [1, 4, 2, 1, 4])
print(vk(100, K) == [2, 1, 4, 2, 1])
print(vk(1000, K) == [1, 4, 2, 1, 4])
vk(10000, K) == [4, 2, 1, 4, 2]
```

## <center>Exercice sur les tuples</center>

Un tuple est une structure permettant de stocker des éléments hétérogènes :

```python
tuple0 = (0, 5.4, "chaîne", True)
tuple0
```

En revanche, les tuples ne sont pas mutables. Il est impossible de modifier un élément, d'en rajouter un ou d'en supprimer un sans redéfinir tout le tuple. L'avantage d'un tuple sur une liste est qu'il est hashable, ce qui implique qu'il peut être inclus dans un ensemble ou utilisé comme clé dans un dictionnaire. Une fonction de hashage est une fonction injective permettant de calculer un identifiant unique pour chaque entrée, et donc de vérifier si deux entrées sont identiques en comparant leurs identifiants, ce qui est beaucoup plus rapide que de comparer toutes les entrées. Une fonction de hashage ne peut s'appliquer que sur des éléments immutables car sinon il faudrait modifier l'identifiant à chaque modification de l'élément. Enfin, il est possible d'affecter des variables à l'intérieur d'un tuple, par exemple :

```python
(a, b) = (1, 9)
(a, b)
```

Cela est en particulier très utile pour échanger deux variables sans avoir à utiliser une variable supplémentaire :

```python
(a, b) = (b, a)
(a, b)
```

### Question 2 : Définir une fonction avec une liste de tuples

```python
def mdlast(lst, val):
    """Inputs: lst <-- List of integer tuples
               val <-- Integer
       Output: lst <-- List of integer tuples with the last element of each tuple replaced by val"""
    for _ in range(len(lst)):
        lst[_] = lst[_][:len(lst[_]) - 1] + (val,)
    return lst
lst = [(10, 20), (30, 40, 50, 60), (70, 80, 90)]
mdlast(lst, 100)
```

## <center>Exercice sur les ensembles</center>

Un ensemble est une structure permettant de stocker des éléments hétérogènes au sens mathématique de la théorie des ensembles :

```python
set0 = {0, 5.4, "chaîne", True}
set0
```

Il est possible de tester si un élément appartient à un ensemble :

```python
if "chaîne" in set0:
    print("Le mot 'chaîne' appartient à l'ensemble.")
```

Les ensembles sont mutables, il est donc possible de rajouter ou de retirer un élément d'un ensemble :

```python
set0.add(18)
set0.add(0)
set0.remove("chaîne")
set0
```

En revanche, les ensembles ne peuvent contenir que des éléments hashables c'est-à-dire immutables. La raison est que les ensembles utilisent des tables de hashage permettant de calculer très rapidement si un élément est présent dans un ensemble, bien plus rapidement que dans une liste ou un tuple. En particulier, un ensemble ne peut pas contenir un autre ensemble :

```python
# set1 = {{1, 2}, {3}, {4}}
# TypeError: unhashable type: 'set'
```

Notons qu'il existe également en Python des ensembles immutables **frozenset** :

```python
frozenset0 = frozenset([0, 5.4, "chaîne", True])
frozenset0
```

Une chaîne de caractères peut être transformée en ensemble :

```python
set1 = set('abracadabra')
set1
```

Comme pour les listes, il est possible de faire des compréhensions d'ensembles :

```python
set2 = {x for x in 'abracadabra' if x not in 'abc'}
set2
```

Dans cet exemple, les chaînes de caractères sont automatiquement transformées en ensemble. Il faut noter que l'ensemble vide est défini par **set()**.

### Question 1 : Ecrire une fonction l'ensemble des nombres entiers divisibles par n inférieurs ou égaux à 100

```python
def divisible(n):
    """Return the set of integers divisible by n less than or equal to 100"""
    return {i for i in range(1, 101) if i % n == 0}
divisible(11)
```

## <center>Exercice sur les dictionnaires</center>

Un dictionnaire est une structure permettant de stocker des éléments hétérogènes indexés par des clés, elles-mêmes hétérogènes :

```python
dict0 = {"pommes": 0, "poires": 4, 12:2}
dict0
```

Les éléments d'un dictionnaire sont accessibles par les clés :

```python
dict0["poires"]
```

Un dictionnaire peut être vu comme un tableau associatif qui à chaque clé associe une valeur. La liste des clés et celle des valeurs sont accessibles respectivement avec **dict0.keys()** et **dict0.values()**. Les dictionnaires sont mutables, il est donc possible de modifier une association clé-valeur, d'en rajouter une ou d'en supprimer une :

```python
dict0["pommes"] = 3
dict0["oranges"] = "beaucoup"
del dict0["poires"]
dict0.pop("pommes")
dict0
```

Bien qu'un dictionnaire soit mutable, les clés qui le composent doivent être des objets hashables c'est-à-dire immutables. La raison est que Python utilise une fonction de hashage sur les clés pour des questions de performance. Ainsi, un ensemble ou une liste ne peuvent pas servir de clés dans un dictionnaire :

```python
# dict0[list0] = "test"
# TypeError: unhashable type: 'list'
# dict0[set0] = "retest"
# TypeError: unhashable type: 'set'
```

En revanche, il est possible d'avoir un tuple ou un frozenset comme clé :

```python
dict0[tuple0] = "test"
dict0
```

```python
dict0[frozenset0] = "rest"
dict0
```

d'où l'intérêt des frozensets. Comme pour les ensembles et les listes, il est possible de faire des compréhensions de dictionnaires :

```python
dict1 = {x: x**2 for x in range(5)}
dict1
```

Finalement, une chose intéressante avec les dictionnaires est l'unpacking illustré ci-dessous :

```python
def add(a = 0, b = 0):
    return a + b
d = {'a': 2, 'b': 3}
add(**d)
```

### Question 4 : Revoir la question précédente avec un dictionnaire pour une évaluation rapide

```python
def build_mots_dict(mots):
    mots_dict = {}
    for mot in mots:
        for n, x in enumerate(mot):
            key = (x, n)
            if key not in mots_dict:
                mots_dict[key] = []
            mots_dict[key].append(mot)
    return mots_dict
mots_dict = build_mots_dict(mots)
mots_dict['e', 4]
```
